```python
#pie

subject=['math','physic','cs','english']
scores=[88,66,48,66]
plt.pie(scores,labels=subject,autopct='%.1f%%',shadow=True,startangle=0)
#autopct are show prenectage ; .1 means how much number after precentage;
plt.title("pie chart")
plt.show()
```

```python
#box
batch1=np.random.randint(10,30,50);
batch2=np.random.randint(10,30,50);
batch3=np.random.randint(10,30,50);
batch4=np.random.randint(10,30,50);
data=[batch1,batch2,batch3,batch4]
# plt.boxplot(batch1,label=batch1)
plt.boxplot(data,tick_labels=['batch1','batch2','batch3','batch4'])

plt.title("box plot")
plt.ylabel("scores")
plt.tight_layout()
plt.show()
```

```python
df=pd.read_csv('PANDASss.csv')
df.head()
```

```python
sns.lineplot(data=df,x='study_hours',y='exam_score')
```

```python
def grade(scores):
  if scores>=80:
    return 'A'
  elif scores>=70:
    return 'B'
  elif scores>=60:
    return 'C'
  elif scores>=50:
    return 'D'
  else:
    return 'F'
df['grade']=df['exam_score'].apply(grade)
df.head()
```

```python
sns.barplot(data=df,x='grade',y='exam_score',estimator='mean')
# ///median ,sum,max,min;
plt.title("performance")


plt.show()
```
