```python
class person:
  def __init__(self,name,age,gender):
    self.name=name
    self.age=age
    self.gender=gender
class rahul(person):
  def __init__(self,name,age,gender):

    super().__init__(name,age,gender)
    # super init are used to as a pass a value;
    self.name=name
    self.age=age
    self.gender=gender

a=rahul('rahul',19,'male')


print(a.gender)
```

```python
class Complex:
    def __init__(self, r, i):
        self.real = r
        self.imag = i

    def show(self):
        print(self.real, "+", self.imag, "i")


c1 = Complex(2, 3)
c2 = Complex(4, 5)

print("First Complex Number:")
c1.show()

print("Second Complex Number:")
c2.show()

print("Sum:")
print(c1.real + c2.real, "+", c1.imag + c2.imag, "i")
```
